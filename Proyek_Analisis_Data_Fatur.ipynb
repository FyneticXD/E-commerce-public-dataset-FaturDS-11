## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana perkembangan jumlah pesanan unik dan total revenue Olist dari bulan ke bulan selama Januari 2017 hingga Agustus 2018?
- **Pertanyaan 2:** Kategori produk apa yang menghasilkan revenue tertinggi dan memiliki jumlah produk terjual terbanyak pada periode analisis?
- **Pertanyaan 3:** Bagaimana karakteristik pelanggan dan distribusi segmen pelanggan berdasarkan RFM (Recency, Frequency, Monetary)?

### Cleaning Data

#### Fixing ... problem

```python

# ==========================================
# 4. CLEANING DATA
# Memperbaiki tipe data kolom tanggal
# ==========================================

# Kolom yang akan dikonversi menjadi datetime
datetime_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

# Mengubah tipe data kolom tanggal
for column in datetime_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

# Mengecek tipe data setelah konversi
print("Tipe Data Setelah Cleaning:")
print(orders[datetime_columns].dtypes)

# Mengecek missing values setelah cleaning
print("\nMissing Values Setelah Cleaning:")
print(orders[datetime_columns].isnull().sum())

# Menampilkan 5 data pertama
orders.head()
```

```python

# ==========================================
# 5. EXPLORATORY DATA ANALYSIS (EDA)
# Eksplorasi awal dataset
# ==========================================

# 1. Distribusi status pesanan
print("Distribusi Status Pesanan:")
print(orders["order_status"].value_counts())

# 2. Statistik deskriptif harga produk
print("\nStatistik Deskriptif Harga Produk:")
print(order_items["price"].describe())

# 3. Tren jumlah pesanan per bulan
monthly_orders = (
    orders.groupby(
        orders["order_purchase_timestamp"].dt.to_period("M")
    )["order_id"]
    .count()
)

print("\nJumlah Pesanan per Bulan:")
print(monthly_orders)

# 4. Menggabungkan order_items dengan products
product_sales = order_items.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

# 5. Revenue dan jumlah produk terjual per kategori
category_sales = (
    product_sales.groupby("product_category_name")
    .agg(
        total_revenue=("price", "sum"),
        total_items_sold=("order_item_id", "count")
    )
    .sort_values("total_revenue", ascending=False)
)

print("\n10 Kategori Produk dengan Revenue Tertinggi:")
print(category_sales.head(10))

# 6. Menampilkan 10 kategori dengan jumlah produk terjual tertinggi
print("\n10 Kategori Produk dengan Penjualan Tertinggi:")
print(
    category_sales.sort_values(
        "total_items_sold",
        ascending=False
    ).head(10)
)
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python

# ==========================================
# PERTANYAAN 1
# Tren Pesanan dan Revenue Bulanan
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# Menghitung revenue dari setiap pesanan
order_revenue = order_items.groupby("order_id")["price"].sum().reset_index()

# Menggabungkan data orders dengan revenue
monthly_data = orders.merge(order_revenue, on="order_id", how="left")

# Mengubah kolom tanggal menjadi datetime
monthly_data["order_purchase_timestamp"] = pd.to_datetime(
    monthly_data["order_purchase_timestamp"]
)

# Mengambil data Januari 2017 - Agustus 2018
monthly_data = monthly_data[
    (monthly_data["order_purchase_timestamp"] >= "2017-01-01") &
    (monthly_data["order_purchase_timestamp"] < "2018-09-01")
]

# Mengelompokkan data berdasarkan bulan
monthly_data["month"] = monthly_data[
    "order_purchase_timestamp"
].dt.to_period("M")

monthly_summary = monthly_data.groupby("month").agg(
    total_orders=("order_id", "nunique"),
    total_revenue=("price", "sum")
).reset_index()

monthly_summary["month"] = monthly_summary["month"].astype(str)

# Membuat visualisasi
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Grafik jumlah pesanan
axes[0].plot(
    monthly_summary["month"],
    monthly_summary["total_orders"],
    marker="o"
)
axes[0].set_title("Tren Jumlah Pesanan Bulanan")
axes[0].set_xlabel("Bulan")
axes[0].set_ylabel("Jumlah Pesanan")
axes[0].tick_params(axis="x", rotation=45)
axes[0].grid(True)

# Grafik revenue
axes[1].plot(
    monthly_summary["month"],
    monthly_summary["total_revenue"],
    marker="o"
)
axes[1].set_title("Tren Revenue Bulanan")
axes[1].set_xlabel("Bulan")
axes[1].set_ylabel("Revenue (BRL)")
axes[1].tick_params(axis="x", rotation=45)
axes[1].grid(True)

plt.tight_layout()
plt.show()

# Menampilkan hasil tabel
monthly_summary
```

### Pertanyaan 2:

```python

# ==========================================
# PERTANYAAN 2
# Kategori Produk dengan Revenue Tertinggi
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# Menggabungkan data order_items dengan products
category_data = order_items.merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

# Menghitung revenue setiap kategori
category_revenue = category_data.groupby(
    "product_category_name"
).agg(
    total_revenue=("price", "sum"),
    total_items_sold=("order_item_id", "count")
).reset_index()

# Mengurutkan berdasarkan revenue tertinggi
top_10_categories = category_revenue.sort_values(
    by="total_revenue",
    ascending=False
).head(10)

# Membuat visualisasi
plt.figure(figsize=(12, 7))

plt.barh(
    top_10_categories["product_category_name"],
    top_10_categories["total_revenue"]
)

plt.title("Top 10 Kategori Produk Berdasarkan Revenue")
plt.xlabel("Total Revenue (BRL)")
plt.ylabel("Kategori Produk")

plt.gca().invert_yaxis()
plt.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()

# Menampilkan tabel hasil
top_10_categories
```

**Insight:** (Opsional)
- Kategori produk beleza_saude menghasilkan revenue tertinggi sebesar 1.258.681,34 BRL.
- Kategori relogios_presentes menempati posisi kedua dengan revenue sebesar 1.205.805,68 BRL.
- Kategori cama_mesa_banho memiliki jumlah item terjual tertinggi, yaitu sebanyak 11.115 item.
- Kategori produk dengan revenue tertinggi menunjukkan potensi yang besar untuk mendukung peningkatan pendapatan bisnis.

## Analisis Lanjutan (Opsional)

```python

# ==========================================
# ANALISIS LANJUTAN
# Analisis Tren Revenue Bulanan
# ==========================================

import matplotlib.pyplot as plt

# Menggabungkan data orders dan order_items
orders_items = orders.merge(
    order_items,
    on="order_id",
    how="inner"
)

# Mengambil data revenue dan jumlah pesanan per bulan
monthly_sales = orders_items.groupby(
    orders_items["order_purchase_timestamp"].dt.to_period("M")
).agg(
    total_revenue=("price", "sum"),
    total_orders=("order_id", "nunique")
).reset_index()

# Mengubah periode menjadi format datetime
monthly_sales["order_purchase_timestamp"] = (
    monthly_sales["order_purchase_timestamp"].astype(str)
)

# Menampilkan hasil
print("Tren Revenue Bulanan:")
print(monthly_sales)

# Visualisasi tren revenue bulanan
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["order_purchase_timestamp"],
    monthly_sales["total_revenue"],
    marker="o"
)

plt.title("Tren Revenue Bulanan E-Commerce")
plt.xlabel("Bulan")
plt.ylabel("Total Revenue (BRL)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Revenue bulanan menunjukkan tren peningkatan dari tahun 2016 hingga 2017, yang mengindikasikan adanya pertumbuhan aktivitas penjualan.
- Pendapatan mengalami fluktuasi pada beberapa bulan, tetapi secara umum menunjukkan tren yang positif hingga tahun 2018.
- Penurunan revenue yang sangat drastis pada September 2018 kemungkinan disebabkan oleh data yang belum lengkap, sehingga perlu diperhatikan dalam interpretasi hasil analisis.
